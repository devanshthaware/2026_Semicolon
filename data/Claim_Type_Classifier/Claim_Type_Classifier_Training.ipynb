# Claim Type Classifier Training

```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import joblib

df=pd.read_csv("claim_type_classifier_dataset_10000.csv")
X_train,X_test,y_train,y_test=train_test_split(df["claim"],df["claim_type"],test_size=0.2,random_state=42)

pipe=Pipeline([
("tfidf",TfidfVectorizer()),
("clf",LogisticRegression(max_iter=1000))
])
pipe.fit(X_train,y_train)
pred=pipe.predict(X_test)
print(classification_report(y_test,pred))
joblib.dump(pipe,"claim_type_classifier.pkl")
```

```
sample=["The Earth revolves around the Sun."]
print(pipe.predict(sample))
```
