# Trust Calibration (Conformal)

```
import pandas as pd
from sklearn.isotonic import IsotonicRegression
import joblib

df=pd.read_csv("trust_calibration_dataset_10000.csv")
iso=IsotonicRegression(out_of_bounds="clip")
iso.fit(df["raw_trust_score"],df["ground_truth"])
df["calibrated"]=iso.predict(df["raw_trust_score"])
print(df.head())
joblib.dump(iso,"trust_calibration.pkl")
```
